```python
import pandas as pd
from sqlalchemy import create_engine

def get_engine(user, password, host, db):
    # The connection string (URL) follows this format:
    # dialect+driver://username:password@host:port/database
    connection_url = f"mysql+mysqlconnector://{user}:{password}@{host}/{db}"
    
    # create_engine returns an Engine object that pandas understands
    engine = create_engine(connection_url)
    return engine

# usage
engine = get_engine("root", "harshit", "localhost", "job_trends")
data = pd.read_sql_table("raw_job_data", engine)
```
